We assume that:

$$
X_i \sim \mathbb{N}(0,1) \text{ for } i = 1, ..., N
$$

## Sum of squares

\begin{align}
\sum X_i^2 &= \frac{1}{N} (\sum X_i)^2 + U_2 \\
U_2 &\sim \chi^2(N-1) \\
\sum X_i &\sim \mathbb{N}( 0, N )
\end{align}

## Sum of products

$$
\sum X_i Y_i = \frac{1}{N}(\sum_i X_i)(\sum_i Y_i) + \frac{1}{2} U_1 - \frac{1}{2} U_2
$$

$$
U_1, U_2 \sim \chi^2(n-1)
$$

```python
# Parameters (you can change N or reps if you like)
N = 10           # number of variables Xi ~ N(0,1); assume i = 1,...,N
reps = 200_000   # number of Monte Carlo repetitions

# Simulate
X = rng.normal(0.0, 1.0, size=(reps, N))
Y = rng.normal(0.0, 1.0, size=(reps, N))
T = (X*Y).sum(axis=1)

# calculate with summary statistics
sum1 = np.random.normal(0, np.sqrt(N), reps)
sum2 = np.random.normal(0, np.sqrt(N), reps)
U1 = np.random.chisquare(N-1, reps)
U2 = np.random.chisquare(N-1, reps)
Tprime = (sum1*sum2) / N + (U1 - U2) / 2
```

```python
fig2 = plt.figure()
plt.hist(Tprime, bins=120, density=True, alpha=0.6)
plt.hist(T, bins=120, density=True, alpha=0.6)
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```
